In this tutorial,  you will apply an encoder-decoder architecture to the problem of 2D (or 3D) human pose estimation. Let's first focus on 2D pose estimation. Then, if you have time, you can extend it to 3D.

In deep learning, before starting the implementation of the neural network, you need to decide of the following;

**1. Input and output:** This is probably half of the solution:
- What is the input to the network and what is the output,i.e., what do  you want the network to produce as output at runtime, once you have deployed it.
- How would you represent the input an the output.

**2. Network architecture:** You have to figure out what is the best network architecture for the problem at hand. In particualr, you need to decide what are the main blocks (e.g., encoder-decoder, GAN, LSTM?) and what is their role. You will then need to decide the detailed architecture of each block. Often, you will need to tweak it depending on the performance.

**3. Training:** You need to decide:
- Which training data you would use. Do you need to collect and label them yourself or publicly available data sets will be sufficient?
- The loss function you will use during training.

**4. Evaluation and testing:** This is as important as the other steps. You particularly need to figure out what test data you would use and how you would measure the performance/accuracy of the network.
